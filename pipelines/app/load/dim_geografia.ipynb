```python
%sql

CREATE OR REPLACE TEMPORARY VIEW tvw_dim_geografia                              AS
SELECT
    ciudad                                                                      AS CIUDAD
    ,depto                                                                      AS DEPARTAMENTO
    ,CURRENT_DATE()                                                             AS _FECHA_CARGA
    ,'silver.cleaned.tb_sucursales_red'                                         AS _FUENTE
FROM silver.cleaned.tb_sucursales_red;
```

```python
%sql
CREATE TABLE IF NOT EXISTS gold.financiero.dim_geografia (

    ciudad                              STRING
    ,depto                              STRING
    ,_FECHA_CARGA                       DATE
    ,_FUENTE                            STRING

)
USING DELTA
LOCATION 'abfss://gold@stdataknowdeveastus001.dfs.core.windows.net/financiero/dim_geografia';
```

```python
%sql
DELETE FROM gold.financiero.dim_geografia;

INSERT INTO gold.financiero.dim_geografia
SELECT * FROM tvw_dim_geografia;
```
