```python
%sql
CREATE OR REPLACE TEMPORARY VIEW tvw_dim_clientes AS
SELECT 
    id_cli                              AS ID_CLIENTE
    ,CONCAT(nomb_cli, " ", apell_cli)   AS NOMBRE_COMPLETO
    ,tip_doc                            AS TIPO_DOCUMENTO
    ,num_doc                            AS NUMERO_DOCUMENTO
    ,fec_nac                            AS FECHA_NACIMIENTO
    ,FLOOR(MONTHS_BETWEEN(
        CURRENT_DATE(), 
            fec_nac) / 12)              AS EDAD
    ,fec_alta                           AS FECHA_ALTA
    ,cod_segmento                       AS COD_SEGMENTO
    ,score_buro                         AS SCORE_BURO
    ,ciudad_res                         AS CIUDAD_RESIDENCIA
    ,depto_res                          AS DEPARTAMENTO_RESIDENCIA
    ,pais                               AS PAIS_RESIDENCIA
    ,estado_cli                         AS ESTADO_CLIENTE
    ,canal_adqis                        AS CANAL_ADQUISICION
    ,CURRENT_DATE()                     AS _FECHA_CARGA
    ,'silver.cleaned.tb_clientes_core'  AS _FUENTE
FROM silver.cleaned.tb_clientes_core;
```

```python
%sql
CREATE TABLE IF NOT EXISTS gold.financiero.dim_clientes (
    ID_CLIENTE               STRING,
    NOMBRE_COMPLETO          STRING,
    TIPO_DOCUMENTO           STRING,
    NUMERO_DOCUMENTO         STRING,
    FECHA_NACIMIENTO         DATE,
    EDAD                     LONG,
    FECHA_ALTA               DATE,
    COD_SEGMENTO             STRING,
    SCORE_BURO               LONG,
    CIUDAD_RESIDENCIA        STRING,
    DEPARTAMENTO_RESIDENCIA  STRING,
    PAIS_RESIDENCIA          STRING,
    ESTADO_CLIENTE           STRING,
    CANAL_ADQUISICION        STRING,
    _FECHA_CARGA             DATE,
    _FUENTE                  STRING
)
USING DELTA
LOCATION 'abfss://gold@stdataknowdeveastus001.dfs.core.windows.net/financiero/dim_clientes';
```
