```python
%sql
-- VW 2: Resumen mensual de cartera por bucket de mora
-- Responde: exposicion y provision requerida por nivel de riesgo

CREATE OR REPLACE VIEW gold.financiero.vw_cartera_mora AS
SELECT
     PERIODO
    ,BUCKET_MORA
    ,CLASIFICACION_REGULATORIA
    ,COUNT(ID_OBLIGACION)                    AS NUM_OBLIGACIONES
    ,COUNT(DISTINCT ID_CLIENTE)              AS NUM_CLIENTES
    ,ROUND(SUM(SALDO_CAPITAL), 2)            AS SALDO_CAPITAL_TOTAL
    ,ROUND(SUM(SALDO_INTERES), 2)            AS SALDO_INTERES_TOTAL
    ,ROUND(SUM(SALDO_TOTAL), 2)              AS SALDO_TOTAL
    ,ROUND(SUM(PROVISION_ESTIMADA), 2)       AS PROVISION_TOTAL
    ,ROUND(
        SUM(PROVISION_ESTIMADA) /
        NULLIF(SUM(SALDO_CAPITAL), 0) * 100
    , 2)                                     AS PCT_COBERTURA
FROM gold.financiero.fact_cartera
GROUP BY PERIODO, BUCKET_MORA, CLASIFICACION_REGULATORIA;
```
