```python
%sql
-- VW 1: Resumen mensual de transacciones por segmento
-- Responde: volumen e ingresos por segmento de cliente y mes

CREATE OR REPLACE VIEW gold.financiero.vw_transacciones_segmento AS
SELECT
     c.COD_SEGMENTO                          AS SEGMENTO
    ,t.PERIODO                               AS PERIODO
    ,COUNT(t.ID_MOVIMIENTO)                  AS NUM_TRANSACCIONES
    ,SUM(t.MONTO_COP)                        AS MONTO_TOTAL_COP
    ,ROUND(AVG(t.MONTO_COP), 2)              AS MONTO_PROMEDIO_COP
    ,SUM(CASE WHEN t.FLAG_ANOMALIA = 1
              THEN 1 ELSE 0 END)             AS NUM_TRANSACCIONES_SOSPECHOSAS
    ,COUNT(DISTINCT t.ID_CLIENTE)            AS NUM_CLIENTES_ACTIVOS
FROM gold.financiero.fact_transacciones t
INNER JOIN gold.financiero.dim_clientes c
    ON t.ID_CLIENTE = c.ID_CLIENTE
GROUP BY c.COD_SEGMENTO, t.PERIODO;
```
