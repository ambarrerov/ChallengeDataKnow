```python
%sql
-- VW 3: Rentabilidad mensual por producto y canal
-- Responde: que productos y canales generan mas ingresos

CREATE OR REPLACE VIEW gold.financiero.vw_rentabilidad_producto_canal AS
SELECT
     t.PERIODO
    ,p.DESCRIPCION_PRODUCTO
    ,p.FAMILIA_PRODUCTO
    ,t.CODIGO_CANAL
    ,COUNT(DISTINCT t.ID_CLIENTE)            AS NUM_CLIENTES
    ,COUNT(t.ID_MOVIMIENTO)                  AS NUM_TRANSACCIONES
    ,ROUND(SUM(t.MONTO_COP), 2)              AS INGRESO_BRUTO_COP
    ,ROUND(AVG(r.INGRESO_TOTAL), 2)          AS INGRESO_PROMEDIO_CLIENTE
    ,ROUND(AVG(r.CLTV_12M), 2)               AS CLTV_PROMEDIO
FROM gold.financiero.fact_transacciones t
INNER JOIN gold.financiero.dim_productos p
    ON t.CODIGO_PRODUCTO = p.CODIGO_PRODUCTO
LEFT JOIN gold.financiero.fact_rentabilidad_cliente r
    ON  t.ID_CLIENTE = r.ID_CLIENTE
    AND t.PERIODO    = r.PERIODO
GROUP BY t.PERIODO, p.DESCRIPCION_PRODUCTO, p.FAMILIA_PRODUCTO, t.CODIGO_CANAL;
```
